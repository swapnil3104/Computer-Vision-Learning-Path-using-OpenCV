```python
# ============================================
# Module 16 - Object Detection Basics
# Face Detection + Eye Detection + Smile Detection
# ============================================

import cv2
import matplotlib.pyplot as plt

# --------------------------------------------
# Load Haar Cascade Classifiers
# --------------------------------------------

face_cascade = cv2.CascadeClassifier(
    "haarcascades/haarcascade_frontalface_default.xml"
)

eye_cascade = cv2.CascadeClassifier(
    "haarcascades/haarcascade_eye.xml"
)

smile_cascade = cv2.CascadeClassifier(
    "haarcascades/haarcascade_smile.xml"
)

# --------------------------------------------
# Read Image
# --------------------------------------------

image = cv2.imread("Images/person.jpg")

# Convert BGR to RGB for plotting
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# --------------------------------------------
# Detect Faces
# --------------------------------------------

faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(50, 50)
)

print("Faces Detected:", len(faces))

# --------------------------------------------
# Detect Eyes and Smile inside each face
# --------------------------------------------

for (x, y, w, h) in faces:

    # Draw Face Rectangle (Blue)
    cv2.rectangle(
        image_rgb,
        (x, y),
        (x + w, y + h),
        (0, 0, 255),
        3
    )

    # Face ROI
    gray_face = gray[y:y+h, x:x+w]
    color_face = image_rgb[y:y+h, x:x+w]

    # ----------------------------------------
    # Eye Detection
    # ----------------------------------------

    eyes = eye_cascade.detectMultiScale(
        gray_face,
        scaleFactor=1.1,
        minNeighbors=10
    )

    print("Eyes:", len(eyes))

    for (ex, ey, ew, eh) in eyes:

        cv2.rectangle(
            color_face,
            (ex, ey),
            (ex + ew, ey + eh),
            (0, 255, 0),
            2
        )

    # ----------------------------------------
    # Smile Detection
    # ----------------------------------------

    smiles = smile_cascade.detectMultiScale(
        gray_face,
        scaleFactor=1.7,
        minNeighbors=20
    )

    print("Smiles:", len(smiles))

    for (sx, sy, sw, sh) in smiles:

        cv2.rectangle(
            color_face,
            (sx, sy),
            (sx + sw, sy + sh),
            (255, 255, 0),
            2
        )

# --------------------------------------------
# Display Results
# --------------------------------------------

plt.figure(figsize=(8,8))
plt.imshow(image_rgb)
plt.title("Face + Eye + Smile Detection")
plt.axis("off")
plt.show()
```
