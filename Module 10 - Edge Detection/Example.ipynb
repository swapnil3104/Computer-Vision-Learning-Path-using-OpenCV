````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (16,10)
````

````python
image = cv2.imread(r"E:\Open CV\Module 10 - Edge Detection\Images\sample.jpg")

if image is None:
    print("Error: Image not found!")
else:
    print("Image Loaded Successfully!")

# Convert to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
````

````python
plt.figure(figsize=(6,6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
````

### Sobel Edge Detection

#### Parameters

```
cv2.Sobel()

Parameters:

src
ddepth
dx
dy
ksize
```

````python
# Sobel X
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)

# Sobel Y
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

# Gradient Magnitude
sobel = cv2.magnitude(sobel_x, sobel_y)

# Convert to uint8
sobel = cv2.convertScaleAbs(sobel)

plt.figure(figsize=(8,6))
plt.imshow(sobel, cmap='gray')
plt.title("Sobel Edge Detection")
plt.axis("off")
plt.show()
````

### Scharr Edge Detection

````python
# Scharr X
scharr_x = cv2.Scharr(gray, cv2.CV_64F, 1, 0)

# Scharr Y
scharr_y = cv2.Scharr(gray, cv2.CV_64F, 0, 1)

# Gradient Magnitude
scharr = cv2.magnitude(scharr_x, scharr_y)

# Convert to uint8
scharr = cv2.convertScaleAbs(scharr)

plt.figure(figsize=(8,6))
plt.imshow(scharr, cmap='gray')
plt.title("Scharr Edge Detection")
plt.axis("off")
plt.show()
````

### Canny Edge Detection

````python
# Apply Gaussian Blur
blur = cv2.GaussianBlur(gray, (5,5), 0)

# Canny
canny = cv2.Canny(blur, 100, 200)

plt.figure(figsize=(8,6))
plt.imshow(canny, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")
plt.show()
````

````python
cv2.imwrite("sobel_output.jpg", sobel)
cv2.imwrite("scharr_output.jpg", scharr)
cv2.imwrite("laplacian_output.jpg", laplacian)
cv2.imwrite("canny_output.jpg", canny)

print("All edge detection images saved successfully!")
````
