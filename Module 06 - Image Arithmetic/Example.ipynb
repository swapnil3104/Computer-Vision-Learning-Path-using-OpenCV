### Why Should Both Images Have the Same Shape for Arithmetic Operations?

Image arithmetic operations in OpenCV, such as addition, subtraction, multiplication, division, and blending, are performed **pixel by pixel**. Each pixel in the first image is combined with the corresponding pixel in the second image.

For these operations to work correctly, both images must have the **same dimensions (width and height)** and the **same number of color channels**.

### Why Is the Same Shape Required?

Suppose we have two images:

- **Image 1:** 600 × 800 × 3
- **Image 2:** 600 × 800 × 3

Each pixel in Image 1 has a matching pixel in Image 2, allowing OpenCV to perform arithmetic operations successfully.

For example:

```text
Result Pixel = Image1 Pixel + Image2 Pixel
```

However, if the image sizes are different:

- **Image 1:** 600 × 800 × 3
- **Image 2:** 400 × 500 × 3

there is no corresponding pixel for many locations. Because of this mismatch, OpenCV cannot perform the operation and raises an error.

### Shape Requirements

Both images should have the same:

- Width
- Height
- Number of channels (Grayscale = 1, Color = 3)

Example:

```text
Image 1 Shape : (600, 800, 3)
Image 2 Shape : (600, 800, 3)
```

✔ Arithmetic operations can be performed.

```text
Image 1 Shape : (600, 800, 3)
Image 2 Shape : (400, 500, 3)
```

✘ Arithmetic operations cannot be performed.

### Solution

If the images have different sizes, resize one image to match the dimensions of the other before performing arithmetic operations.

```python
import cv2

image1 = cv2.imread("Images/Image1.jpg")
image2 = cv2.imread("Images/Image2.jpg")

# Resize image2 to match image1
image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

print(image1.shape)
print(image2.shape)
```

### Key Points

- Image arithmetic is performed pixel by pixel.
- Both images must have the same width and height.
- Both images must have the same number of channels.
- If the image sizes differ, resize one image before performing arithmetic operations.
- Matching image shapes ensures accurate and error-free image processing.

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# Function to display images
# -------------------------------
def show(title, image):
    if len(image.shape) == 3:
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(5,4))
    plt.imshow(image, cmap="gray")
    plt.title(title)
    plt.axis("off")
    plt.show()


# -------------------------------
# Read Images
# -------------------------------
image1 = cv2.imread("Images/Image1.jpg")
image2 = cv2.imread("Images/Image2.jpg")

# Resize image2 if dimensions differ
image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

# -------------------------------
# 1. Original Images
# -------------------------------
show("Image 1", image1)
show("Image 2", image2)

# -------------------------------
# 2. Image Addition
# -------------------------------
addition = cv2.add(image1, image2)
show("Addition", addition)

# -------------------------------
# 3. Image Subtraction
# -------------------------------
subtraction = cv2.subtract(image1, image2)
show("Subtraction", subtraction)

# -------------------------------
# 4. Increase Brightness
# -------------------------------
brightness = np.full(image1.shape, 50, dtype=np.uint8)
bright = cv2.add(image1, brightness)
show("Brightness +50", bright)

# -------------------------------
# 5. Decrease Brightness
# -------------------------------
dark = cv2.subtract(image1, brightness)
show("Brightness -50", dark)

# -------------------------------
# 6. Bitwise AND
# -------------------------------
bit_and = cv2.bitwise_and(image1, image2)
show("Bitwise AND", bit_and)

# -------------------------------
# 7. Bitwise OR
# -------------------------------
bit_or = cv2.bitwise_or(image1, image2)
show("Bitwise OR", bit_or)

# -------------------------------
# 8. Bitwise XOR
# -------------------------------
bit_xor = cv2.bitwise_xor(image1, image2)
show("Bitwise XOR", bit_xor)

# -------------------------------
# 9. Bitwise NOT
# -------------------------------
bit_not = cv2.bitwise_not(image1)
show("Bitwise NOT", bit_not)

# -------------------------------
# 10. Image Blending (70% + 30%)
# -------------------------------
blend = cv2.addWeighted(image1, 0.7, image2, 0.3, 0)
show("Image Blending (70:30)", blend)

# -------------------------------
# 11. Transparency Overlay (50%)
# -------------------------------
transparent = cv2.addWeighted(image1, 0.5, image2, 0.5, 0)
show("Transparency (50%)", transparent)

# -------------------------------
# 12. Create a Mask
# -------------------------------
mask = np.zeros(image1.shape[:2], dtype=np.uint8)

cv2.rectangle(mask, (100,100), (350,350), 255, -1)

show("Mask", mask)

# -------------------------------
# 13. Apply Mask
# -------------------------------
masked = cv2.bitwise_and(image1, image1, mask=mask)
show("Masked Image", masked)

# -------------------------------
# 14. Watermark Effect
# -------------------------------
watermark = cv2.addWeighted(image1, 0.9, image2, 0.1, 0)
show("Watermark Effect", watermark)

# -------------------------------
# 15. NumPy vs OpenCV Addition
# -------------------------------
a = np.uint8([250])
b = np.uint8([20])

print("NumPy Addition :", a + b)
print("OpenCV Addition:", cv2.add(a, b))
````
