```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

# Read Image
image = cv2.imread("Images/car.jpg")

if image is None:
    print("Error: Image not found!")
else:
    print("Image Loaded Successfully")

# Convert BGR to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

### Average Blur

```python
average_blur = cv2.blur(image, (7,7))

# (7,7) represents the kernel size for the average blur
# The kernel size determines the area of the image that will be averaged to produce the blur effect. 
# A larger kernel size results in a more pronounced blur, while a smaller kernel size retains more detail. 
# In this case, a (7,7) kernel means that each pixel in the output image is computed as the average of a 7x7 neighborhood of pixels from the input image.

plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(average_blur, cv2.COLOR_BGR2RGB))
plt.title("Average Blur")
plt.axis("off")
plt.show()
```

### Gaussian Blur

```python
gaussian_blur = cv2.GaussianBlur(image, (7,7), 0)

# The (7,7) kernel size for Gaussian blur works similarly to the average blur, but it applies a Gaussian function to weight the pixels in the neighborhood.
# The third parameter (0) is the standard deviation in the X direction. If it is set to 0, OpenCV calculates it based on the kernel size. 
# A Gaussian blur tends to produce a smoother and more natural-looking blur compared to average blur, especially for images with varying intensity levels.

plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(gaussian_blur, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Blur")
plt.axis("off")
plt.show()
```

### Median Blur

```python
median_blur = cv2.medianBlur(image, 7)

# The (7) parameter represents the kernel size for the median blur.
# The kernel size must be a positive odd integer. 
# In this case, a kernel size of 7 means that each pixel in the output image is computed as the median of a 7x7 neighborhood of pixels from the input image.

plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(median_blur, cv2.COLOR_BGR2RGB))
plt.title("Median Blur")
plt.axis("off")
plt.show()
```

### Bilateral Filter

```python
bilateral = cv2.bilateralFilter(image, 9, 75, 75)

# (9, 75, 75) are the parameters for the bilateral filter.
# The first parameter (9) is the diameter of the pixel neighborhood used during filtering.
# The second parameter (75) is the filter sigma in the color space. A larger value means that farther colors within the pixel neighborhood will be mixed together, resulting in larger areas of semi-equal color.
# The third parameter (75) is the filter sigma in the coordinate space. A larger value means that farther pixels will influence each other as long as their colors are close enough.

plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB))
plt.title("Bilateral Filter")
plt.axis("off")
plt.show()
```

### Sharpening

```python
kernel = np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
])
# Define the kernel for sharpening
# The kernel is a 3x3 matrix that emphasizes the center pixel while subtracting the neighboring pixels.
# The values in the kernel determine how much influence each pixel has on the output.
# The center value (5) amplifies the intensity of the center pixel, while the surrounding values (-1) reduce the influence of the neighboring pixels, resulting in a sharpened image.

sharpen = cv2.filter2D(image, -1, kernel)

plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(sharpen, cv2.COLOR_BGR2RGB))
plt.title("Sharpened Image")
plt.axis("off")
plt.show()
```

### Compare All Filters


```python
images = [
    image_rgb,
    cv2.cvtColor(average_blur, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(gaussian_blur, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(median_blur, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(sharpen, cv2.COLOR_BGR2RGB)
]

titles = [
    "Original",
    "Average Blur",
    "Gaussian Blur",
    "Median Blur",
    "Bilateral Filter",
    "Sharpening"
]

plt.figure(figsize=(18,10))

for i in range(len(images)):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i])
    plt.title(titles[i], fontsize=12)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
cv2.imwrite("average_blur.jpg", average_blur)
cv2.imwrite("gaussian_blur.jpg", gaussian_blur)
cv2.imwrite("median_blur.jpg", median_blur)
cv2.imwrite("bilateral_filter.jpg", bilateral)
cv2.imwrite("sharpen.jpg", sharpen)

print("All images saved successfully.")
```
